### Exercise 8.1: Learning Curves

Visualize learning curves to diagnose overfitting:
1. Generate learning curves
2. Identify overfitting/underfitting
3. Understand bias-variance trade-off

```python
from sklearn.model_selection import learning_curve

# Your code here
def plot_learning_curve(estimator, X, y, title):
    train_sizes, train_scores, val_scores = learning_curve(
        estimator, X, y, cv=5, n_jobs=-1,
        train_sizes=np.linspace(0.1, 1.0, 10),
        scoring='accuracy'
    )
    
    train_mean = np.mean(train_scores, axis=1)
    train_std = np.std(train_scores, axis=1)
    val_mean = np.mean(val_scores, axis=1)
    val_std = np.std(val_scores, axis=1)
    
    plt.figure(figsize=(10, 6))
    plt.plot(train_sizes, train_mean, label='Training score', linewidth=2)
    plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.2)
    plt.plot(train_sizes, val_mean, label='Validation score', linewidth=2)
    plt.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.2)
    plt.xlabel('Training Set Size')
    plt.ylabel('Accuracy')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

# Compare models
# Overfitting model (high complexity)
overfit_model = DecisionTreeClassifier(max_depth=None, random_state=42)
plot_learning_curve(overfit_model, X, y, 'Learning Curve - High Complexity Tree (Overfitting)')

# Good model (moderate complexity)
good_model = DecisionTreeClassifier(max_depth=3, random_state=42)
plot_learning_curve(good_model, X, y, 'Learning Curve - Moderate Complexity Tree')
```
